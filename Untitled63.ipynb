```python
# Задача 1. Напишите функцию revenue(Q), вычисляющую выручку по формуле R(Q) = p*Q, где цена единицы товара p = 95 руб.
# Напишите функцию cost(Q), вычисляющую издержки по формуле C(Q) = a*Q + b, где переменные издержки на единицу a = 53 руб., постоянные издержки b = 1950 руб.
# Используя функции revenue и cost, напишите функцию profit(Q) = revenue(Q) - cost(Q) и вычислите значение прибыли при объёме продаж Q = 240 единиц.
# Функции должны принимать параметры Q (обязательный) и p, a, b со значениями по умолчанию, равными указанным выше.

# 1. Блок ввода данных (исходные параметры для расчета)
p = 95
a = 53
b = 1950
Q = 240

# 2. Определение функций с вашими значениями по умолчанию
def revenue(Q, p=95):
    """Вычисление выручки"""
    return p * Q

def cost(Q, a=53, b=1950):
    """Вычисление издержек"""
    return a * Q + b

def profit(Q, p=95, a=53, b=1950):
    """Вычисление прибыли как разности выручки и издержек"""
    return revenue(Q, p) - cost(Q, a, b)

# 3. Вызовы функций и вывод результатов
print(f"Выручка: {revenue(Q, p)} руб.")
print(f"Издержки: {cost(Q, a, b)} руб.")
print(f"Прибыль при объёме продаж Q = {Q} ед.: {profit(Q, p, a, b)} руб.")

```

```python
# Задача 2. Цена товара до скидки: P = 280 руб. Размер сезонной скидки: d = 27%.
# С помощью lambda выражения запишите функцию final_price(P, d), вычисляющую итоговую цену после применения скидки: final_price = P * (1 - d/100).
# Вычислите итоговую цену для заданных P и d.

# Исходные данные
P = 280  # Базовая цена
d = 27   # Скидка в %

# Лямбда-функция для расчета цены со СКИДКОЙ (минус вместо плюса)
final_price_func = lambda P, d: P * (1 - d / 100)

# Вычисление итоговой цены
final_price = final_price_func(P, d)

print(f"Базовая цена: {P} руб.")
print(f"Скидка: {d}%")
print(f"Итоговая цена: {final_price} руб.")
```

```python
# Задача 3. Рост выручки компании описывается зависимостью growth(x) = 100*(1 + 4.0/100)**x (x — номер года).
# Напишите универсальную функцию apply_to_all(f, values), которая принимает функцию f и список значений values,
# и возвращает список результатов применения f к каждому значению (аналог встроенной функции map).
# Примените apply_to_all к функции growth и списку годов values = [6, 12, 15, 17, 21].

# 1. Исходный список годов из условия задачи
values = [6, 12, 15, 17, 21]

# 2. Функция роста выручки компании
def growth(x):
    """Экспоненциальная зависимость роста выручки от номера года"""
    return 100 * (1 + 4.0 / 100) ** x

# 3. Универсальная функция-аналог map
def apply_to_all(f, values):
    """Применяет переданную функцию f к каждому элементу списка values"""
    result = []
    for item in values:
        result.append(f(item))  # Берем год, отправляем в growth(x) и сохраняем итог
    return result

# 4. Вызов универсальной функции для расчета выручки по годам
growth_results = apply_to_all(growth, values)

# 5. Вывод результатов на экран
print(f"Исходные года: {values}")
print(f"Результаты роста выручки: {growth_results}")

```

```python
# Задача 4. Реализуйте функцию-декоратор count_calls(func), которая подсчитывает и печатает количество вызовов декорируемой функции.
# Примените декоратор к функции profit(Q) из Задачи 1 (не изменяя её тело — присвойте profit = count_calls(profit))
# и вызовите декорированную функцию три раза подряд — при объёмах продаж 230, 240 и 250 единиц.
#После каждого вызова выведите текущее количество вызовов функции и полученное значение прибыли.

# 1. Три последовательных вызова функции для объемов из ВАШЕГО условия
test_volumes = [230, 240, 250]

# 2. Базовые функции из ВАШЕГО варианта Задачи 1 (p=95, a=53, b=1950)
def revenue(Q, p=95):
    return p * Q

def cost(Q, a=53, b=1950):
    return a * Q + b

def profit(Q, p=95, a=53, b=1950):
    return revenue(Q, p) - cost(Q, a, b)


# 3. Реализация функции-декоратора
def count_calls(func):
    """Декоратор, подсчитывающий число вызовов функции"""
    def wrapper(*args, **kwargs):
        wrapper.calls += 1  # Увеличиваем счетчик при каждом обращении к функции
        return func(*args, **kwargs)
    wrapper.calls = 0  # Начальное значение счетчика вызовов
    return wrapper


# 4. Применение декоратора к profit БЕЗ изменения её тела (через присваивание)
profit = count_calls(profit)


# 5. Цикл для вычислений и вывода результатов
for Q in test_volumes:
    current_profit = profit(Q)
    # Выводим текущее состояние счетчика из wrapper.calls и результат
    print(f"Текущее количество вызовов: {profit.calls}")
    print(f"Объем продаж Q = {Q} ед. -> Прибыль = {current_profit} руб.")
    print("-" * 40)


```

```python
# Задача 5. Прибыль предприятия описывается квадратичной функцией profit(Q) = -1*Q**2 + 140*Q - 1875 (Q — объём продаж, ед.).
# Не используя циклы и итерационные численные методы, найдите аналитически: а)
# точки безубыточности — корни уравнения profit(Q) = 0 (используйте дискриминант и формулу корней квадратного уравнения средствами модуля math;
# предусмотрите обработку случая, когда действительных корней нет); б) объём продаж Q*, при котором прибыль максимальна (по формуле вершины параболы Q* = 140/(2*1)),
# и соответствующее значение максимальной прибыли profit(Q*). Сравните найденную точку максимума с серединой отрезка [0, 20].

import math

# Коэффициенты квадратного уравнения прибыли: -1*Q**2 + 140*Q - 1875 = 0
a = -1
b = 140
c = -1875

print("а) Поиск точек безубыточности (корней уравнения):")
# Вычисляем дискриминант
D = b**2 - 4 * a * c
print(f"Дискриминант D = {D}")

if D > 0:
    # Два корня
    Q1 = (-b + math.sqrt(D)) / (2 * a)
    Q2 = (-b - math.sqrt(D)) / (2 * a)
    # Сортируем корни от меньшего к большему для наглядности вывода
    print(f"Точки безубыточности: Q1 = {min(Q1, Q2):.1f} ед., Q2 = {max(Q1, Q2):.1f} ед.")
elif D == 0:
    # Один корень
    Q0 = -b / (2 * a)
    print(f"Единственная точка безубыточности: Q = {Q0:.1f} ед.")
else:
    # Действительных корней нет
    print("Действительных корней нет. Выход на безубыточность невозможен.")

print("\nб) Поиск точки максимума прибыли:")
# Объём продаж для максимума прибыли по формуле вершины параболы из условия задачи
Q_star = 140 / (2 * 1)

# Расчет максимального значения прибыли подстановкой Q_star в исходную формулу
max_profit = -1 * Q_star**2 + 140 * Q_star - 1875
print(f"Оптимальный объем продаж Q* = {Q_star:.1f} ед.")
print(f"Максимальная прибыль profit(Q*) = {max_profit:.1f} руб.")

# Нахождение середины отрезка [0, 20]
midpoint = (0 + 20) / 2
print(f"Середина отрезка: {midpoint:.1f}")

# Сравнение точки максимума с серединой отрезка
if Q_star > midpoint:
    print(f"Вывод: Точка максимума Q* ({Q_star:.1f}) БОЛЬШЕ середины отрезка ({midpoint:.1f}) на {Q_star - midpoint:.1f} ед.")
elif Q_star < midpoint:
    print(f"Вывод: Точка максимума Q* ({Q_star:.1f}) МЕНЬШЕ середины отрезка ({midpoint:.1f}) на {midpoint - Q_star:.1f} ед.")
else:
    print("Вывод: Точка максимума Q* строго совпадает с серединой отрезка.")

```

```python
# Задача 6. Дан список финансовых результатов подразделений компании за отчётный период (тыс. руб.): results = [750, 0, -400, -650, 1350, 300, -450].
# Напишите обычную функцию classify_result(x), которая возвращает 1, если x > 0 (прибыль), -1, если x < 0 (убыток), и 0, если x == 0 (безубыточно).
# С помощью np.vectorize примените эту функцию поэлементно к массиву NumPy, построенному из списка results, и выведите массив меток.

import numpy as np

# Исходный список финансовых результатов из текста условия
results = [750, 0, -400, -650, 1350, 300, -450]


def classify_result(x):
    """Обычная функция для классификации одного числа"""
    if x > 0:
        return 1  # Прибыль
    elif x < 0:
        return -1  # Убыток
    else:
        return 0  # Безубыточно


# Строим массив NumPy из исходного списка
results_array = np.array(results)

# С помощью np.vectorize делаем функцию пригодной для обработки целых массивов
vectorized_classify = np.vectorize(classify_result)

# Поэлементно применяем функцию к массиву NumPy
labels_array = vectorized_classify(results_array)

print(f"Исходный массив данных: {results_array}")
print(f"Полученный массив меток: {labels_array}")

```

```python
# Задача 7. Дан массив NumPy цен товаров prices = [930, 920, 135, 695, 80, 880, 365, 715, 630] (руб.).
# Выведите: а) элемент с индексом -2; б) срез массива с 1-го по предпоследний элемент;
# в) массив, отсортированный по возрастанию; г) каждый третий элемент массива.

import numpy as np

# Исходный массив цен товаров
prices = np.array([930, 920, 135, 695, 80, 880, 365, 715, 630])

# а) Элемент с индексом -2
elem_minus_2 = prices[-2]

# б) Срез с 1-го (индекс 1) по предпоследний элемент
# В Python правая граница среза не включается, поэтому -1 указывает на предпоследний
slice_middle = prices[1:-1]

# в) Массив, отсортированный по возрастанию
# np.sort() создает новый отсортированный массив, не меняя оригинал
sorted_prices = np.sort(prices)

# г) Каждый третий элемент массива
# Срез со значением шага ::3 берет элементы с индексами 0, 3, 6...
every_third = prices[::3]

# Вывод результатов
print(f"Исходный массив цен: {prices}\n")
print(f"а) Элемент с индексом -2: {elem_minus_2}")
print(f"б) Срез с 1-го по предпоследний элемент: {slice_middle}")
print(f"в) Отсортированный массив: {sorted_prices}")
print(f"г) Каждый третий элемент: {every_third}")
```

```python
# Задача 8. Дан массив NumPy доходностей акции за 9 торговых дней (в %): returns = [-8.0, 5.1, -2.3, 12.5, -10.1, 4.7, -8.9, 12.8, -10.3].
# а) с помощью булевой индексации выведите только положительные значения доходности; б) выведите количество дней, когда доходность была ниже -8.0%;
# в) создайте новый массив, в котором все значения доходности ниже -8.0% заменены на 0
# (отрицательные результаты ниже порога считаются некорректными данными и обнуляются).

import numpy as np

# Исходный массив доходностей акции за 9 дней
returns = np.array([-8.0, 5.1, -2.3, 12.5, -10.1, 4.7, -8.9, 12.8, -10.3])

# а) Булевая индексация: только положительные значения (> 0)
positive_returns = returns[returns > 0]

# б) Количество дней, когда доходность была ниже -8.0%
days_count = np.sum(returns < -8.0)

# в) Создание нового массива с заменой значений ниже -8.0% на 0
cleaned_returns = np.where(returns < -8.0, 0.0, returns)

# Вывод результатов в точном соответствии с вашим скриншотом
print(f"Исходный массив доходностей: {returns}\n")
print(f"а) Только положительные доходности: {positive_returns}")
print(f"б) Количество дней с доходностью ниже -8.0%: {days_count}")
print(f"в) Массив с обнуленными некорректными данными: {cleaned_returns}")


```

```python
# Задача 9. Дана матрица NumPy объёмов продаж 3 товаров за 4 месяца (ед.): sales = [[90, 210, 285, 240], [60, 25, 20, 180], [245, 25, 215, 190]] (строки — товары, столбцы — месяцы).
# Вычислите: а) суммарный объём продаж по каждому товару (сумма по строкам); б) суммарный объём продаж по каждому месяцу (сумма по столбцам);
# в) среднее значение продаж по каждому месяцу (среднее по Практическая работа № 1 · Функции и NumPy в экономике стр. 6 из 66 столбцам);
# г) с помощью broadcasting постройте матрицу отклонений продаж от среднемесячного значения (вычтите из каждого столбца его среднее, посчитанное в пункте в)).

import numpy as np

# Исходная матрица продаж (строки — товары, столбцы — месяцы)
sales = np.array([[90, 210, 285, 240], [60, 25, 20, 180], [245, 25, 215, 190]])

# а) Суммарный объем продаж по каждому товару (сумма по СТРОКАМ)
# axis=1 означает схлопывание столбцов, то есть подсчет вдоль строки
total_per_product = np.sum(sales, axis=1)

# б) Суммарный объем продаж по каждому месяцу (сумма по СТОЛБЦАМ)
# axis=0 означает схлопывание строк, то есть подсчет по вертикали
total_per_month = np.sum(sales, axis=0)

# в) Среднее значение продаж по каждому месяцу (среднее по СТОЛБЦАМ)
mean_per_month = np.mean(sales, axis=0)

# г) Матрица отклонений от среднемесячного значения (broadcasting)
# NumPy автоматически вычтет одномерный массив средних из каждого столбца матрицы
deviations = sales - mean_per_month

# Вывод результатов
print("Исходная матрица продаж:\n", sales, "\n")
print(f"а) Суммарные продажи по товарам (по строкам): {total_per_product}")
print(f"б) Суммарные продажи по месяцам (по столбцам): {total_per_month}")
print(f"в) Средние продажи по месяцам (по столбцам): {mean_per_month}\n")
print("г) Матрица отклонений от среднего:\n", deviations)

```

```python
# Задача 10. Равновесие на рынке двух взаимосвязанных товаров описывается системой линейных уравнений: 1.58*x1 + (-1.85)*x2 = 50; 1.98*x1 + 2.19*x2 = 145,
# где x1, x2 — равновесные объёмы товаров.
# Средствами numpy.linalg.solve найдите x1 и x2.
# Проверьте правильность решения, вычислив A @ x и сравнив с вектором правой части с помощью numpy.allclose.

import numpy as np

# 1. Задаем матрицу коэффициентов A и вектор свободных членов B
A = np.array([[1.58, -1.85], [1.98, 2.19]])
B = np.array([50, 145])

# 2. Находим решение системы (равновесные объемы x1 и x2)
# Функция np.linalg.solve находит вектор x, при котором A @ x = B
x = np.linalg.solve(A, B)

# 3. Проверка правильности решения
# Вычисляем произведение матрицы A на полученный вектор x
check_B = A @ x

# Сравниваем полученный вектор с исходным вектором B (с учетом погрешности float)
is_correct = np.allclose(check_B, B)

# Вывод результатов в консоль
print(f"Матрица коэффициентов A:\n{A}\n")
print(f"Вектор правой части B: {B}\n")
print(f"Равновесные объемы товаров (x1, x2): {x}")
print(f"Результат умножения A @ x: {check_B}")
print(f"Проверка успешна (совпадает с B)? {is_correct}")
```
